```python
# Importaciones de librerías
import matplotlib.pyplot as plt  # Librería para visualización
from sklearn.cluster import KMeans  # Algoritmo de clustering
from sklearn.datasets import make_blobs  # Dataset para generación de los datos

# Generación de datos
X, _ = make_blobs(n_samples=300, centers=4, cluster_std=2.0, random_state=0)
print(X.shape)
```

En la generación de datos solo almacenamos X, ya que contiene las coordenadas de los datos que serán utilizadas por KMeans para generar los clusters. El segundo valor devuelto por la función se almacena como `_` porque no será utilizado en este desarrollo.


### Creamos un objeto de KMeans con Clusters


```python
kmeans = KMeans(n_clusters=4)  # Instanciación del modelo KMeans con 4 clusters
kmeans.fit(X)  # Ajuste del modelo a los datos generados
labels = kmeans.labels_  # Etiquetas de cluster asignadas a cada punto de datos
centers = kmeans.cluster_centers_  # Coordenadas de los centros de los clusters
print(centers)  # Visualización de los clusters y sus centros
```

Con KMeans y el método `fit()`, se ajusta el modelo a los datos generados. Luego, se obtienen las etiquetas de los clusters asignados a cada punto de datos y las coordenadas de los centroides mediante `cluster_centers_`. Estas coordenadas representan el centro de cada grupo generado por el algoritmo. Por ejemplo, si un centroide se encuentra en las coordenadas (1.4, 0.2), significa que dicho punto representa la posición central de uno de los clusters identificados.


```python
# Visualización de los clusters y sus centros
plt.scatter(
    X[:, 0], X[:, 1], c=labels, cmap="viridis", marker="o", edgecolor="k", s=100
)  # Puntos de datos coloreados por cluster
plt.scatter(
    centers[:, 0], centers[:, 1], c="red", marker="X", s=200, label="Centroides"
)  # Centroides de los clusters
plt.title("K-Means Clustering")  # Título del gráfico
plt.xlabel("Característica 1")  # Etiqueta del eje X
plt.ylabel("Característica 2")  # Etiqueta del eje Y
plt.legend()  # Leyenda para identificar los centroides
plt.show()  # Mostrar el gráfico con los clusters y sus centros
```

En la visualización podemos observar cómo, mediante el dataset generado con `make_blobs`, se crean 300 registros distribuidos en 4 centros. Posteriormente, utilizando KMeans, los datos son agrupados en 4 clusters. A través de la variable `labels`, cada registro recibe una etiqueta que indica el grupo al que pertenece, permitiendo visualizar los distintos clusters mediante colores. Además, la marca "X" representa los centroides obtenidos por KMeans, los cuales corresponden al punto central de cada grupo y sirven como referencia para la agrupación de los datos.
